# 📘 NumPy Quick Reference Guide

A concise cheat sheet of commonly used NumPy functions for quick revision and explanation.

---

## 🔹 Array Operations
- `np.sort()` – Sorts elements of an array in ascending order  
- `np.append()` – Adds values to the end of an array  
- `np.concatenate()` – Joins two or more arrays along a given axis  
- `np.unique()` – Returns sorted unique elements  

---

## 🔹 Shape Manipulation
- `np.expand_dims()` – Expands array shape by adding a new axis  
- `np.flip()` – Reverses the order of elements  

---

## 🔹 Conditional & Indexing
- `np.where()` – Returns indices where condition is true  
- `np.argmax()` – Returns index of maximum value  
- `np.isin()` – Checks if elements are present in another array  

---

## 🔹 Mathematical & Statistical Functions
- `np.cumsum()` – Computes cumulative sum  
- `np.percentile()` – Computes specified percentile  
- `np.histogram()` – Computes frequency distribution  
- `np.corrcoef()` – Computes correlation coefficient matrix  

---

## 🔹 Array Modification
- `np.put()` – Replaces elements at specified indices  
- `np.delete()` – Removes elements from array  
- `np.clip()` – Limits values within a specified range  

---

## 🔹 Set Operations
- `np.union1d()` – Union of two arrays (unique values)  
- `np.intersect1d()` – Intersection (common elements)  
- `np.setdiff1d()` – Difference (elements in first not in second)  
- `np.setxor1d()` – Symmetric difference (non-common elements)  
- `np.in1d()` – Tests whether each element is in another array

- **Set Ops** → `np.union1d()`, `np.intersect1d()`, `np.setdiff1d()`, `np.setxor1d()`
