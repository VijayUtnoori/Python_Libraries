Numpy array VS python lists

```python
# speed ,list
a=[i for i in range(10000000)]
b=[i for i in range(10000000,20000000)]
c=[]
import time
start=time.time()
for i in range(len(a)):
  c.append(a[i]+b[i])
print(time.time()-start)
```

```python
#numpy
import numpy as np
a=np.arange(10000000)
b=np.arange(10000000,20000000)
start=time.time()
c=a+b
print(time.time()-start)
```

```python
1.6791675090789795/0.19632768630981445 # numpy is 8 times faster than python list
```

Broadcasting

the term broadcasting tells about how numpy treat array with different shapes during arithmetic operation

Broadcasting Rules

1)Make dimensions equal

If one array has fewer dimensions, just imagine adding 1 in front of it until both have the same number of dimensions.

2)Match sizes

If sizes don’t match:

2.1)If one of them is 1, it can expand (copy itself) to match the other.

2.2)If neither is 1, then ❌ error (broadcasting not possible).

Screenshot 2026-04-25 151902.png

Broadcasting Rule (simple)

Two arrays can work if:

Same shape ✅
OR
One dimension is 1 ✅

```python
# same shape
a=np.arange(6).reshape(2,3)
b=np.arange(6,12).reshape(2,3)

print(a)
print(b)
print(a+b)
```

```python
#2D=>(3,2) ,1D=>(3) ==(1,3) after convert the same size
#3D=>(3,3,3),1D=>(3)==(1,1,3) ''
```

```python
a=np.arange(6).reshape(2,3)
b=np.arange(3)
print(a)
print(b)
print("adition of a+b",a+b)
```

```python
# brpadcast not run here because the shape is not match
a=np.arange(12).reshape(3,4)
b=np.arange(3)
print(a)
print("b array",b)
print(a+b)
```

```python
a=np.arange(3).reshape(1,3)
b=np.arange(3).reshape(3,1)
print(a)
print(b)
print("addition of a+b",a+b)
```

```python
a=np.arange(3).reshape(1,3)
b=np.arange(4).reshape(4,1)
print("addition of a+b ",a+b)
```

```python
#can not make shape equal
a=np.arange(12).reshape(4,3)
b=np.arange(12).reshape(3,4)
print(a)
print(b)
print("addition of a+b,",a+b)
```
